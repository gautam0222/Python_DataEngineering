```python
from pydantic import BaseModel
```

```python
class employee_schema(BaseModel):
    name: str
    id: int
    department: str
```

```python
obj = employee_schema(name="John Doe", id=123, department="HR")
obj
```

```python
obj2 = employee_schema(**{"name": "John Doe", "id": 123, "department": "HR"})
obj2
#Packing and Unpacking of dictionary in python
```

## Coercion VS Strict Types

```python
obj3 = employee_schema(**{"name": "John Doe", "id": 101, "department": "HR"})
obj3
#coerce the data type of id from str to int but it will not coerce the data type of name from int to str because it is not possible to convert int to str
```

```python
from pydantic import StrictInt

class employee_schema(BaseModel):
    name: str
    id: StrictInt
    department: str

obj4 = employee_schema(**{"name": "John Doe", "id": 101, "department": "HR"})
obj4
#This will raise an error because the id is a strict int, it is a string.
```
