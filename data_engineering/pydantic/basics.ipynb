```python
from pydantic import BaseModel
```

```python
class employee_schema(BaseModel):
    name: str
    id: int
    department: str
```

```python
obj = employee_schema(name="John Doe", id=123, department="HR")
obj
```

```python
obj2 = employee_schema(**{"name": "John Doe", "id": 123, "department": "HR"})
obj2
#Packing and Unpacking of dictionary in python
```

## Coercion VS Strict Types

```python
obj3 = employee_schema(**{"name": "John Doe", "id": "101", "department": "HR"})
obj3
# The non-strict int field accepts and converts this numeric string.
```

```python
from pydantic import StrictInt, ValidationError

class employee_schema(BaseModel):
    name: str
    id: StrictInt
    department: str

try:
    employee_schema(**{"name": "John Doe", "id": "101", "department": "HR"})
except ValidationError as error:
    print(error)  # A numeric string is rejected by StrictInt.
```
