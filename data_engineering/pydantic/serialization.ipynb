```python
from pydantic import BaseModel, Field, EmailStr
from typing import Optional, Literal

class employee_schema(BaseModel):
    name: str = Field(..., min_length=1, max_length=50)
    id: int = Field(..., strict=True)
    department: Optional[str] = Field(None, max_length=50)
    age: int = Field(..., ge=18, le=65)
    email: EmailStr = Field(..., description="Employee's email address")
    gender: Literal['Male', 'Female', 'Other'] = Field(..., description="Employee's gender")
```

```python
obj1 = employee_schema(**{"name": "John Doe", "id": 123, "department": "HR", "age": 30, "email": "John.Doe@example.com", "gender": "Male"})
obj1
```

```python
obj1.model_dump()  # Convert the Pydantic model to a dictionary
```

```python
print(obj1.model_dump().keys())
print(obj1.model_dump().values())
print(obj1.model_dump().items())
print(obj1.model_dump_json())  # Convert the Pydantic model to a JSON string
```
